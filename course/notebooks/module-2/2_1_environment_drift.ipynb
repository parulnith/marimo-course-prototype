# The "It Works on My Machine" Problem

This notebook demonstrates **environmental drift** when the same code produces different results because the environment changed.

Use it by running the same notebook in two different environments manually and comparing the results.

## Example 1: scikit-learn 1.8 feature change

This example uses a feature that exists in scikit-learn 1.8 but not in 1.7.2: `method="temperature"` in `CalibratedClassifierCV`.

Run the cell below in one environment, note what happens, then run the same cell again in the other environment and compare the result.

- In scikit-learn 1.8.0, it should run.
- In scikit-learn 1.7.2, it should fail because `temperature` calibration is not available.

```python
import sklearn
from sklearn.calibration import CalibratedClassifierCV
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

print(f"sklearn version: {sklearn.__version__}")

X, y = make_classification(n_samples=300, n_features=10, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

base_model = LogisticRegression(max_iter=1000)
model = CalibratedClassifierCV(base_model, method="temperature", cv=3)
model.fit(X_train, y_train)
accuracy = model.score(X_test, y_test)

print(f"accuracy: {accuracy:.3f}")
```
